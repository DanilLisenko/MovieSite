```python
# ============================================================
# ПОИСК ЧАСТЫХ НАБОРОВ
# ============================================================

import warnings
warnings.filterwarnings("ignore")

!pip install mlxtend -q

import pandas as pd
import matplotlib.pyplot as plt
import time

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori

# ============================================================
# ЗАГРУЗКА ДАННЫХ
# ============================================================
df = pd.read_csv(
    "baskets.csv",
    encoding="cp1251"
)

print("=" * 60)
print("РАЗМЕР ДАННЫХ")
print("=" * 60)

print(df.shape)

print("\nПервые строки:\n")

display(df.head())



# ============================================================
# СОЗДАНИЕ ТРАНЗАКЦИЙ
# ============================================================

transactions = []

for _, row in df.iterrows():

    # Удаляем пустые значения
    items = row.dropna().tolist()

    transactions.append(items)

print("\nПример транзакции:\n")

print(transactions[0])



# ============================================================
# ONE-HOT ENCODING
# ============================================================

# Преобразуем данные в формат:
#
# milk bread butter
# 1      1      0

te = TransactionEncoder()

te_array = te.fit(transactions).transform(transactions)

basket_matrix = pd.DataFrame(
    te_array,
    columns=te.columns_
)

print("\nМатрица корзин:\n")

display(basket_matrix.head())



# ============================================================
# SUPPORT ДЛЯ ЭКСПЕРИМЕНТОВ
# ============================================================

support_values = [0.01, 0.03, 0.05, 0.10, 0.15]



# ============================================================
# СПИСКИ ДЛЯ ГРАФИКОВ
# ============================================================

runtime_results = []

all_length_stats = {}



# ============================================================
# ЗАПУСК APRIORI
# ============================================================

for support in support_values:

    print("\n")
    print("=" * 60)
    print(f"MIN_SUPPORT = {support}")
    print("=" * 60)

    # ========================================================
    # ЗАПУСК АЛГОРИТМА
    # ========================================================

    start_time = time.time()

    frequent_itemsets = apriori(
        basket_matrix,
        min_support=support,
        use_colnames=True
    )

    end_time = time.time()

    runtime = end_time - start_time

    runtime_results.append(runtime)

    # ========================================================
    # ДОБАВЛЯЕМ ДЛИНУ НАБОРОВ
    # ========================================================

    frequent_itemsets["length"] = (
        frequent_itemsets["itemsets"].apply(len)
    )

    # ========================================================
    # СОРТИРОВКА ПО SUPPORT
    # ========================================================

    frequent_itemsets = frequent_itemsets.sort_values(
        by="support",
        ascending=False
    )

    # ========================================================
    # ОБЩАЯ ИНФОРМАЦИЯ
    # ========================================================

    print(f"\nКоличество наборов: {len(frequent_itemsets)}")

    print(f"Время работы: {runtime:.4f} сек")



    # ========================================================
    # ВЫВОД КОЛИЧЕСТВА НАБОРОВ КАЖДОЙ ДЛИНЫ
    # ========================================================

    length_counts = (
        frequent_itemsets["length"].value_counts()
    )

    all_length_stats[support] = length_counts

    print("\nКоличество наборов разных длин:\n")

    for length in sorted(length_counts.index):

        if length == 1:
            name = "Одиночные наборы"

        elif length == 2:
            name = "Пары"

        elif length == 3:
            name = "Тройки"

        elif length == 4:
            name = "Четверки"

        elif length == 5:
            name = "Пятерки"

        else:
            name = f"Наборы длины {length}"

        print(f"{name}: {length_counts[length]}")



    # ========================================================
    # ТАБЛИЦА С ДЛИНАМИ
    # ========================================================

    length_table = pd.DataFrame({

        "Длина набора": length_counts.index,
        "Количество": length_counts.values

    })

    print("\nТаблица длин:\n")

    display(
        length_table.sort_values("Длина набора")
    )



    # ========================================================
    # ВЫВОД ТОП-20 НАБОРОВ
    # ========================================================

    print("\nТОП-20 frequent itemsets:\n")

    display(
        frequent_itemsets.head(20)
    )



# ============================================================
# ГРАФИК 1
# ВРЕМЯ РАБОТЫ
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    [s * 100 for s in support_values],
    runtime_results,
    marker='o'
)

plt.xlabel("Минимальная поддержка (%)")

plt.ylabel("Время работы (сек)")

plt.title(
    "Быстродействие алгоритма Apriori"
)

plt.grid(True)

plt.show()

# ============================================================
# ГРАФИК 2
# КОЛИЧЕСТВО НАБОРОВ РАЗНОЙ ДЛИНЫ
# ============================================================

plt.figure(figsize=(12, 7))

for support, counts in all_length_stats.items():

    plt.plot(
        counts.index,
        counts.values,
        marker='o',
        label=f"{int(support * 100)}%"
    )

plt.xlabel("Длина набора")

plt.ylabel("Количество наборов")

plt.title(
    "Количество frequent itemsets разных длин"
)

plt.legend(title="Support")

plt.grid(True)

plt.show()



# ============================================================
# ЛЕКСИКОГРАФИЧЕСКАЯ СОРТИРОВКА
# ============================================================

print("\n")
print("=" * 60)
print("ЛЕКСИКОГРАФИЧЕСКАЯ СОРТИРОВКА")
print("=" * 60)

frequent_itemsets["itemsets_str"] = (

    frequent_itemsets["itemsets"].apply(

        lambda x: ", ".join(sorted(list(x)))
    )
)

lexicographic_sorted = frequent_itemsets.sort_values(
    by="itemsets_str"
)

display(

    lexicographic_sorted[
        ["support", "itemsets"]
    ].head(20)
)



# ============================================================
# СОХРАНЕНИЕ РЕЗУЛЬТАТОВ
# ============================================================

frequent_itemsets.to_csv(
    "frequent_itemsets.csv",
    index=False
)

print("\nФайл frequent_itemsets.csv сохранен.")
```
